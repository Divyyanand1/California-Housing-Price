## 1. Setup

Import the libraries and load the California Housing dataset from scikit-learn.

```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np
from sklearn.datasets import fetch_california_housing
from sklearn.model_selection import train_test_split

housing = fetch_california_housing()
```

## 2. Load and Explore the Data

### 2.1 Feature Names

```python
housing.feature_names
```

### 2.2 Build the DataFrame

```python
housing_df = pd.DataFrame(housing["data"], columns=housing["feature_names"])
housing_df
```

### 2.3 Add the Target Column

The target is the median house value (in units of $100,000).

```python
housing_df["target"] = housing["target"]
housing_df.head()
```

## 3. Dataset Description

- **Number of instances:** 20,640
- **Number of attributes:** 8 numeric predictive attributes and the target

| Feature | Description |
|---|---|
| `MedInc` | Median income in block group |
| `HouseAge` | Median house age in block group |
| `AveRooms` | Average number of rooms per household |
| `AveBedrms` | Average number of bedrooms per household |
| `Population` | Block group population |
| `AveOccup` | Average number of household members |
| `Latitude` | Block group latitude |
| `Longitude` | Block group longitude |
| `target` | Median house value (in $100,000s) |

## 4. Model Training and Evaluation

Each model below is trained on 80% of the data (20% held out with `train_test_split`). The score reported is the R² returned by `model.score()` on the **training** set.

### 4.1 Define Features and Target

The same split is used by all of the models below.

```python
X = housing_df.drop("target", axis=1)
y = housing_df["target"]

X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2)
```

### 4.2 Ridge Regression

```python
from sklearn.linear_model import Ridge

model = Ridge()
model.fit(X_train, y_train)
model.score(X_train, y_train)
```

**Result:** Ridge regression reached an R² of about 0.60 on the training set.

### 4.3 Linear Regression

```python
from sklearn.linear_model import LinearRegression

model = LinearRegression()
model.fit(X_train, y_train)
model.score(X_train, y_train)
```

**Result:** Linear regression reached an R² of about 0.60 on the training set.

### 4.4 Lasso Regression

```python
from sklearn.linear_model import Lasso

model = Lasso()
model.fit(X_train, y_train)
model.score(X_train, y_train)
```

**Result:** Lasso regression reached an R² of about 0.28 on the training set.

### 4.5 Random Forest Regressor

```python
from sklearn.ensemble import RandomForestRegressor

model = RandomForestRegressor()
model.fit(X_train, y_train)
model.score(X_train, y_train)
```

**Result:** The Random Forest reached an R² of about 0.97 on the training set.

## 5. Results Summary

| Model | Training R² (approx.) |
|---|---|
| Ridge | 0.61 |
| Linear Regression | 0.60 |
| Lasso | 0.28 |
| Random Forest | 0.97 |

> **Note:** These scores are measured on the training data. A very high training score (as with the Random Forest) can indicate overfitting. To compare the models fairly, evaluate them on the held-out test set with `model.score(X_test, y_test)` or use cross-validation. Exact values will vary between runs because the train/test split is random (set `random_state` for reproducible results).
